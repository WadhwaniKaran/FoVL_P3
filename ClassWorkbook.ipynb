# L1

```python
%matplotlib inline
from matplotlib import pyplot as plt
import numpy as np
from pprint import pprint
```

```python
dataX = ['the dog likes the cat',
         'a cat runs in the park',
         'the cat sees a dog in the park',
         'the dog sees a cat',
         'a dog runs in the park',
         'the cat likes the dog',
         'a cat sees the dog',
         'the dog likes milk']

dataLabel = [0, 1, 1, 0, 1, 0, 1, 1]
```

```python
# tokenize

sentences_tok = dataX.copy()

for i in range(len(dataX)):
    sentences_tok[i] = dataX[i].split()
```

```python
# print tokens

for i in sentences_tok:
    print(i)
```

```python
# vocab
vocab = []
for i in range(len(sentences_tok)):
    vocab += set(sentences_tok[i])
vocab = sorted(list(set(vocab)))
```

```python
print(len(vocab))
```

```python
def oneHotEncode(data):
    dataEncoded = []
    for i in range(len(data)):
        vec = [0 for _ in range(len(vocab))]
        for j in range(len(data[i])):
            idx = vocab.index(data[i][j])
            vec[idx] = 1
        dataEncoded.append(vec)
    return dataEncoded
```

```python
oHE = oneHotEncode(sentences_tok)
```

```python
# print one hot encoded labels

for i in oHE:
    print(i)
```

```python
dataSamples = len(dataX)
vocabSize = len(vocab)
```

```python
def weightMat(numNeurons):
    dims = len(vocab) + 1
    mat = np.random.randn(numNeurons, dims)
    return mat
```

```python
inNeurons = 3
mat = weightMat(inNeurons)
print(mat)
```

# L2

```python
rng = np.random.default_rng(seed=42)
```

```python
print(vocab)
```

```python
windowSize = 2
embedDim = 5
```

```python
 wordToID, idToWord = {}, {}

for (idx, word) in enumerate(vocab):
    idToWord[idx] = word
    wordToID[word] = idx
pprint(idToWord)
```

```python
cenCtxPair = [] # center as in and ctx as target

for sent_tok in sentences_tok:
    for idx in range(len(sent_tok)):
        indices = [
            i + idx for i in range(-windowSize, windowSize + 1) 
            if (i + idx) >= 0 and (i + idx) < len(sent_tok)
        ]
        for j in indices:
            if j == idx:
                continue
            cenCtxPair.append([wordToID[sent_tok[j]], wordToID[sent_tok[idx]]])
```

```python
print(sentences_tok[0])
```

```python
print(cenCtxPair)
```

```python
def getRandom2DMat(numRows, numCols):
    return rng.standard_normal((numRows, numCols)) 
```

```python
# Input embedding
E = getRandom2DMat(vocabSize, embedDim) * 0.01
```

```python
# Weight matrix
W = getRandom2DMat(embedDim, vocabSize)
```

```python
print(f"E dims: {E.shape} and W dims: {W.shape}")
```

```python
def softmax2D(x, axis=-1):
    max_x = np.max(x, axis=axis, keepdims=True)
    exp_x = np.exp(x - max_x)
    
    return exp_x / np.sum(exp_x, axis=axis, keepdims=True)
```

```python
def forwardPass(centerWord):
    hiddenVec = E[wordToID[centerWord]]
    scores = hiddenVec @ W
    probs = softmax2D(scores)
    return probs
```

```python
(centerWordIdx, contextWordIdx) = cenCtxPair[0]
centerWord, contextWord = idToWord[contextWordIdx], idToWord[centerWordIdx]
print(centerWord, contextWord)
probs = forwardPass(centerWord)
pred = probs[wordToID[contextWord]]
loss = -np.log(pred)
print(f"Loss: {loss}")
```
