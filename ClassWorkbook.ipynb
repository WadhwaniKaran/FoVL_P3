```python
%matplotlib inline
from matplotlib import pyplot as plt
import numpy as np
```

```python
dataX = ['The dog likes the cat.',
         'A cat runs in the park.',
         'The cat sees a dog in the park.',
         'The dog sees a cat.',
         'A dog runs in the park.',
         'The cat likes the dog.',
         'A cat sees the dog.',
         'The dog likes milk.']

dataLabel = [0, 1, 1, 0, 1, 0, 1, 1]
```

```python
# tokenize

sentences_tok = dataX

for i in range(len(dataX)):
    sentences_tok[i] = dataX[i].split()
```

```python
# print tokens

for i in sentences_tok:
    print(i)
```

```python
# vocab
vocab = []
for i in range(len(sentences_tok)):
    vocab += set(sentences_tok[i])
vocab = list(set(vocab))
```

```python
print(len(vocab))
```

```python
def oneHotEncode(data):
    dataEncoded = []
    for i in range(len(data)):
        vec = [0 for _ in range(len(vocab))]
        for j in range(len(data[i])):
            idx = vocab.index(data[i][j])
            vec[idx] = 1
        dataEncoded.append(vec)
    return dataEncoded
```

```python
oHE = oneHotEncode(sentences_tok)
```

```python
# print one hot encoded labels

for i in oHE:
    print(i)
```

```python
dataSamples = len(dataX)
vocabSize = len(vocab)
```

```python
def weightMat(numNeurons):
    dims = len(vocab) + 1
    mat = np.random.randn(numNeurons, dims)
    return mat
```

```python
inNeurons = 3
mat = weightMat(inNeurons)
print(mat)
```

```python
# pred 
n1 = 3

for i in range(dataSamples):
    x = oHE[i]
    y = dataLabel[i]
    
    mat1 = weightMat(n1)
    
    for j in range(n1):
        _y = 0
        for k in range(vocabSize):
            _y += mat1[j][k] * x[k]
        _y += mat1[j][-1]
    
        print(f"{i} for neuron {j}: wx+b={_y}")
    
```
